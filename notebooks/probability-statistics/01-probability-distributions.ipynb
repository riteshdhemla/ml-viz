# Probability Distributions

Companion notebook for the [Probability Distributions](https://ml-viz.vercel.app/courses/probability-statistics/01-probability-distributions) lesson.

We'll explore key probability distributions used in ML and build intuition for expectation and variance.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.style.use('dark_background')
plt.rcParams.update({
    'figure.facecolor': '#0f1117', 'axes.facecolor': '#1a1d27',
    'axes.edgecolor': '#30344a', 'text.color': '#e2e8f0',
    'axes.labelcolor': '#e2e8f0', 'xtick.color': '#94a3b8', 'ytick.color': '#94a3b8',
})
```

## Key distributions at a glance

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# 1. Bernoulli
ax = axes[0, 0]
for p, color in [(0.3, '#6366f1'), (0.5, '#2dd4bf'), (0.7, '#f97316')]:
    ax.bar([0, 1], [1-p, p], width=0.2, alpha=0.8, color=color,
           label=f'p={p}', align='center')
ax.set_xticks([0, 1]); ax.set_xticklabels(['0 (failure)', '1 (success)'])
ax.set_title('Bernoulli distribution'); ax.legend()

# 2. Gaussian with different parameters
ax = axes[0, 1]
x = np.linspace(-6, 10, 400)
for mu, sigma, color in [(0, 1, '#6366f1'), (3, 1.5, '#f97316'), (0, 2, '#2dd4bf')]:
    ax.plot(x, stats.norm.pdf(x, mu, sigma), color=color, lw=2,
            label=f'N({mu}, {sigma}²)')
ax.set_title('Gaussian distributions'); ax.legend()

# 3. Central Limit Theorem
rng = np.random.default_rng(42)
ax = axes[1, 0]
for n, color in [(1, '#6366f1'), (5, '#f97316'), (30, '#2dd4bf')]:
    means = rng.uniform(0, 1, size=(5000, n)).mean(axis=1)
    ax.hist(means, bins=50, alpha=0.5, density=True, color=color, label=f'n={n}')
ax.set_title('Central Limit Theorem\n(mean of n uniform[0,1] samples)')
ax.legend()

# 4. 2D Gaussian
ax = axes[1, 1]
xy = np.mgrid[-3:3:.05, -3:3:.05].reshape(2, -1).T
for mu, cov, color, label in [
    ([0, 0], [[1, 0.8], [0.8, 1]], '#6366f1', 'ρ=0.8'),
    ([0, 0], [[1, -0.5], [-0.5, 1]], '#f97316', 'ρ=-0.5'),
]:
    density = stats.multivariate_normal(mu, cov).pdf(xy).reshape(120, 120)
    X_ = np.linspace(-3, 3, 120)
    ax.contour(X_, X_, density, levels=5, colors=[color], alpha=0.7)
    ax.plot([], [], color=color, label=label)
ax.set_aspect('equal'); ax.set_title('2D Gaussian with different covariances')
ax.legend()

for ax in axes.flat: ax.grid(True, alpha=0.2)
plt.suptitle('Key Distributions in Machine Learning', y=1.01, fontsize=13)
plt.tight_layout(); plt.show()
```

## Expectation and variance empirically

```python
rng = np.random.default_rng(0)

distributions = [
    ('Bernoulli(0.7)',  rng.binomial(1, 0.7, 10000), 0.7, 0.7*0.3),
    ('Gaussian(2, 4)', rng.normal(2, 2, 10000),       2,   4),
    ('Uniform[0, 1]',  rng.uniform(0, 1, 10000),      0.5, 1/12),
]

print(f'{'Distribution':20s}  {'E[X] (true)':12s}  {'E[X] (sample)':14s}  {'Var (true)':12s}  {'Var (sample)':13s}')
print('-' * 80)
for name, samples, true_mean, true_var in distributions:
    print(f'{name:20s}  {true_mean:12.4f}  {samples.mean():14.4f}  {true_var:12.4f}  {samples.var():13.4f}')
```
